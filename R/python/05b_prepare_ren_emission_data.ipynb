# 05b: Prepare REN emission data

In this notebook, we prepare the REN emissions data.

```python
import pyam
import pandas as pd
from climate_assessment.climate import clean_wg3_scenarios

from pathlib import Path
import copy
```

Step 1: Read in the AR6 data. You should have downloaded a file called `AR6_Scenarios_Database_World_ALL_CLIMATE.csv` and placed it in the Data/pathways folder.

```python
df = pd.read_csv(
    Path(
        #'../../Data/pathways/AR6_Scenarios_Database_World_ALL_CLIMATE_v1.1.csv'
        '/Users/gauravganti/Documents/ca_projects/2023_overshoot_perspective/data/AR6_Scenarios_Database_World_ALL_CLIMATE_v1.1.csv'
    ),
    index_col=[0,1]
)
```

Step 2: Filter out the Ren scenario.

```python
df_imp = df.loc[
    pd.IndexSlice[
        ('REMIND-MAgPIE 2.1-4.3','REMIND-MAgPIE 2.1-4.2'),
        ('DeepElec_SSP2_ HighRE_Budg900', 'SusDev_SDP-PkBudg1000')
    ],
    :
]
```

Step 3: Cast to a pyam dataframe (so we can make use of the filter functions..)

```python
df_imp_pyam = pyam.IamDataFrame(
    df_imp
    .reset_index()
)
```

Step 4: Filter out the necessary variables

```python
df_imp_emissions = (
    df_imp_pyam
    .filter(
        variable='AR6 climate diagnostics|Infilled|Emissions*'
    )
)
```

Step 5: Clean and prepare the scenarios for the climate assessment

```python
df_imp_emissions_cleaned = pyam.IamDataFrame(
    clean_wg3_scenarios(
        df_imp_emissions
    )
)
```

Step 6: Write this out for further assessment.

```python
df_imp_emissions_cleaned.to_csv(
    Path(
        '../../Data/pathways/ar6_imp_rensp/ar6_ren_sp_emissions_for_climate_assessment.csv'
    )
)
```
