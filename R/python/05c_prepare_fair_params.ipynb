# 05c Prepare FaIR parameters

In this notebook, we prepare the FaIR configurations that we will use for the climate assessment.

```python
import pooch
import os
import json
```

Step 1: Define the data paths and filenames

```python
data_dir = os.path.join(
    '..',
    '..',
    'Data',
    'fair'
)
os.makedirs(
    data_dir,
    exist_ok=True
)
```

```python
fair_slim_filename = 'fair-1.6.2-wg3-params-slim.json'
fair_common_filename = 'fair-1.6.2-wg3-params-common.json'
```

Step 2: Retrieve the two parameter sets.

```python
fair_slim_url = (
    "https://zenodo.org/record/6601980/files/fair-1.6.2-wg3-params-slim.json?download=1"
)
fair_slim_hash = "c071ca619c0ae37a6abdeb79c0cece7b"

pooch.retrieve(
    url=fair_slim_url,
    known_hash=f"md5:{fair_slim_hash}",
    path=data_dir,
    fname=fair_slim_filename,
)
```

```python
fair_common_url = "https://zenodo.org/record/6601980/files/fair-1.6.2-wg3-params-common.json?download=1"
fair_common_hash = "42ccaffcd3dea88edfca77da0cd5789b"

pooch.retrieve(
    url=fair_common_url,
    known_hash=f"md5:{fair_common_hash}",
    path=data_dir,
    fname=fair_common_filename,
)
```

Step 3: Read in the configuration with the solar cycle forcing.

```python
with open(os.path.join('../../Data/fair/', fair_common_filename)) as f:
    extra_config = json.load(f)
```

Step 5: Remove solar cycle forcing in the projected period.

```python
extra_config['default_solar'][(2016-1750):] = [0] * 95
```

Step 6: Write out this file.

```python
with open(os.path.join('../../Data/fair/', fair_common_filename), 'w') as f:
    json.dump(extra_config, f)
```
